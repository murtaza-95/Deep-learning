```python
import numpy as np
import keras
```

```python
import tensorflow as tf
```

```python
xs = np.array(object=[-1, 0, 1, 2, 3, 4], dtype=np.float32)
ys = np.array(object=[-3, -1, 1, 3, 5, 7], dtype=np.float32)
```

```python
model = keras.Sequential()  #stack of layers
model.add(layer=keras.Input(shape=(1,)))
model.add(layer=keras.layers.Dense(units=1))
```

```python
model.summary()
```

```python
model.compile(
    optimizer=keras.optimizers.SGD(learning_rate=0.01),
    loss=keras.losses.MeanSquaredError(),
    metrics=[
        keras.metrics.MeanAbsoluteError()
        ],
)
```

```python
history = model.fit(
    x=xs,
    y=ys,
    epochs=500,
    verbose=1,
)
```

```python
y_pred = model.predict(x=np.array(object=[10], dtype=np.float32))
```
